MANIPULATING THE DATASET

1. WHAT IS THE TOTAL REVENUE GENERATED OVER TIME

```python
import pandas as pd
import matplotlib.pyplot as plt

df['Date'] = pd.to_datetime(df['Date'])
rev = df.groupby(df['Date'].dt.date)['Total_Cost'].sum().reset_index()
rev['Date'] = pd.to_datetime(rev['Date'])

plt.figure(figsize=(16,10))
plt.bar(rev['Date'], rev['Total_Cost'], color='skyblue')
plt.xlabel("Date")  
plt.ylabel("Total Revenue")
plt.title("Total Revenue by Date")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

2. Which day of the week has the highest number of transactions?

```python
import matplotlib.pyplot as plt
df['Date'] = pd.to_datetime(df['Date'])
df['Day_of_Week'] = df['Date'].dt.day_name()

transactions_by_day = df['Day_of_Week'].value_counts()

days_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
transactions_by_day = transactions_by_day.reindex(days_order)

plt.figure(figsize=(10, 5))
plt.plot(transactions_by_day.index, transactions_by_day.values, color='skyblue',marker='o')

plt.xlabel("Day of the Week")
plt.ylabel("Number of Transactions")
plt.title("Transactions Per Day of the Week")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

3. CITY  WITH HIGHEST REVENUE

```python
import matplotlib.pyplot as plt

city_revenue = df.groupby('City')['Total_Cost'].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
plt.plot(city_revenue.index, city_revenue.values, color='teal',marker='o')

plt.xlabel("City")
plt.ylabel("Total Revenue ($)")
plt.title("Total Revenue by City")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```


4. What is the average purchase value per transaction?

```python
average_purchase_value = df['Total_Cost'].mean()

print(f"The average purchase value per transaction is ${average_purchase_value:.2f}")
```

5. Which Store Generate Highest Revenue ?

```python
highest_revenue_store=df.groupby('Store_Type')['Total_Cost'].sum()
plt.figure(figsize=(12, 6))

plt.plot(highest_revenue_store.index, highest_revenue_store.values, color='orange',marker='o')
plt.xlabel("Store_Type")

plt.ylabel("Revenue ($)")

plt.title("Revenue by Store Type")

plt.show()
```

6. Which Customer Category Spends the Most ?

9. How many transactions involved discounts?

```python
import matplotlib.pyplot as plt


promotion_counts = df['Promotion'].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(promotion_counts, labels=promotion_counts.index, autopct='%1.1f%%', colors=plt.cm.Pastel1.colors)
plt.title('Distribution of Promotions in Transactions')
plt.show()
```

10. Which promotion strategy performed best in terms of revenue?'

```python
promotion_revenue=df.groupby('Promotion')['Total_Cost'].sum().sort_values(ascending=False)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(promotion_revenue.index,promotion_revenue.values,marker='o',color='coral')

plt.xlabel('Promotion Strategy')
plt.ylabel('Total Revenue')
plt.title('Revenue by Promotion Strategy')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

11. Which season sees the most purchases?

```python
season=df['Season'].value_counts()
seasons_order = ['Spring', 'Summer', 'Fall', 'Winter']
season_counts = season.reindex(seasons_order)
plt.figure(figsize=(8, 5))
plt.plot(season_counts.index, season_counts.values, color='lightseagreen',marker='o')
plt.xlabel('Season')
plt.ylabel('Number of Purchases')
plt.title('Purchases by Season')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```

12. which day of the week had the highest total revenue

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Day_of_Week'] = df['Date'].dt.day_name()

daily_revenue = df.groupby('Day_of_Week')['Total_Cost'].sum()

days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_revenue = daily_revenue.reindex(days_order)

plt.figure(figsize=(10, 6))
plt.plot(daily_revenue.index, daily_revenue.values, color='red',marker='o')
plt.xlabel('Day of the Week')
plt.ylabel('Total Revenue')
plt.title('Total Revenue by Day of the Week')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
